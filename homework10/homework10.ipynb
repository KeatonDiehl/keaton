```python
cols_to_skip = ["Planet", "Ring System?", "Global Magnetic Field?"]
for col in df.columns:
    if col not in cols_to_skip:
        df[col] = pd.to_numeric(df[col], errors='coerce')
df.dtypes
```

```python
# float64: decimal numbers
# int64: whole integers
# object: text strings (i.e. y/n)

# planet, ring system?, and global magnetic field where excluded since their data needs to be a string (planet name or y/n)
```

```python
import astropy.units as u

def attach_units(col_name, unit, new_name):
    if col_name in df.columns:
        new_values = []
        
        for val in df[col_name]:
            new_values.append(val * unit)
            
        df[col_name] = new_values
        
        df.rename(columns={col_name: new_name}, inplace=True)

attach_units("Mass (10^24kg)", 1e24 * u.kg, "Mass (kg)")

attach_units("Diameter (km)", u.km, "Diameter (km)")

attach_units("Density (kg/m3)", u.kg / u.m**3, "Density (kg/m^3)")

attach_units("Gravity (m/s2)", u.m / u.s**2, "Gravity (m/s^2)")

attach_units("Escape Velocity (km/s)", u.km / u.s, "Escape Velocity (km/s)")

attach_units("Rotation Period (hours)", u.hour, "Rotation Period (hours)")

attach_units("Length of Day (hours)", u.hour, "Length of Day (hours)")

attach_units("Distance from Sun (10^6 km)", 1e6 * u.km, "Distance from Sun (km)")

attach_units("Perihelion (10^6 km)", 1e6 * u.km, "Perihelion (km)")

attach_units("Aphelion (10^6 km)", 1e6 * u.km, "Aphelion (km)")

attach_units("Orbital Period (days)", u.day, "Orbital Period (days)")

attach_units("Orbital Velocity (km/s)", u.km / u.s, "Orbital Velocity (km/s)")

attach_units("Orbital Inclination (degrees)", u.deg, "Orbital Inclination (degrees)")

attach_units("Obliquity to Orbit (degrees)", u.deg, "Obliquity to Orbit (degrees)")

attach_units("Mean Temperature (C)", u.deg_C, "Mean Temperature (C)")

attach_units("Surface Pressure (bars)", u.bar, "Surface Pressure (bars)")

df.head(1)
```

```python
semi_major_axis = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2

loc_index = df.columns.get_loc("Aphelion (km)") + 1

df.insert(loc_index, "Semi-Major Axis (km)", semi_major_axis)

print(df.columns.tolist())
```

```python
df["Semi-Major Axis (km)"]
```

```python
df['Orbital Period (days)'] = df['Orbital Period (days)'].apply(lambda x: x.to(u.yr))

df.rename(columns={'Orbital Period (days)': 'Orbital Period (years)'}, inplace=True)

print("Updated Columns:", df.columns)

print(df['Orbital Period (years)'])

fave_planet = df.loc[df['Planet'] == 'Saturn'].iloc[0]
print(f"\nSaturn's Orbital Period: {fave_planet['Orbital Period (years)'].value:.4f} years")
```

```python
print(f"1 AU in Astropy: {const.au}")

km_in_au = const.au.to(u.km)
print(f"Kilometers in 1 AU: {km_in_au}")

cols_to_convert = [col for col in df.columns if '(km)' in col]

print(f"\nConverting the following columns to AU: {cols_to_convert}")

for col in cols_to_convert:
    df[col] = df[col].apply(lambda x: x.to(u.AU))
    
    new_col_name = col.replace('(km)', '(AU)')
    
    df.rename(columns={col: new_col_name}, inplace=True)

saturn_row = df[df['Planet'] == 'Saturn'].iloc[0]

for col in df.columns:
    if '(AU)' in col:
        print(f"{col}: {saturn_row[col]}")
```
