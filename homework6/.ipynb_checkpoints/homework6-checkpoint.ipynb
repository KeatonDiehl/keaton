```python
# 4: Prime Star Systems

import math

arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])

# Use trail division to determine if prime
def is_prime(n):
    # Handle special cases
    if n <= 1:
        return False
    if n <= 3:
        return True
    
    # Handle even numbers
    if n % 2 == 0:
        return False
        
    # Check for factors from 3 up to sqrt(n)
    # The step is 2 to only check odd numbers
    i = 3
    while i <= math.sqrt(n):
        if n % i == 0:
            return False
        i += 2
        return True

def find_primes(arr):
    prime_rows = []
    for row in arr:
           if any(is_prime(num) for num in row):
            prime_rows.append(row)
    return np.array(prime_rows)
print(find_primes(arr))
```

```python
# 5: Checkers

def checkerboard():
    board = np.zeros((7, 7), dtype = int)
    # even rows:
    board[::2, ::2] = 1
    # odd rows:
    board[1::2, 1::2] = 1
    return board
print(checkerboard())
```

```python
# 6: Expanding Universe

universe = np.array(['galaxy', 'clusters'])

def expansion(arr, n):
    expanded_list = []
    spacer = ' ' * n
    for character in universe:
        character_list = list(character)
        expanded_list.append(spacer.join(list(character)))
    return np.array(expanded_list)

print(expansion(universe, 2))
```

```python
# 7: Second-Brightest Star

arr = np.array([[1123, 1456, 1789, 1324, 1876],
                [1567, 1987, 1678, 1405, 1589],
                [1345, 1654, 1523, 1109, 1923],
                [1298, 1890, 1367, 1784, 1432],
                [1823, 1756, 1489, 1672, 1550]])

def second_brightest(arr):
    sorted_stars = np.sort(arr, axis=1)
    return sorted_stars[:, -2]

print(second_brightest(arr))
        
```
