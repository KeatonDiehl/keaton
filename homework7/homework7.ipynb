```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, axes = plt.subplots(1, 3, figsize=(25, 5))
ax1, ax2, ax3 = axes

for i, (A, w) in enumerate(zip(amplitude, frequency)):

    # --- First Subplot: Vary only the amplitude
    w_fixed = 1
    y1 = make_sine_wave(x, A, w_fixed)
    ax1.plot(x, y1,label=f'A={A:.1f}, w={w_fixed}')
    
    # --- Second Subplot: Vary only the frequency
    A_fixed = 1
    y2 = make_sine_wave(x, A_fixed, w)
    ax2.plot(x, y2, label=f'A={A_fixed}, w={w}')

    # --- Third Subplot: Vary amplitude and frequency together
    y3 = make_sine_wave(x, A, w)
    ax3.plot(x, y3, label=f'A={A:.1f}, w={w}')


# Subplot 1: Varying Amplitude
ax1.set_title('Varying Amplitude', fontsize=16)
ax1.set_xlabel('x', fontsize=14)
ax1.set_ylabel('y', fontsize=14)
ax1.legend(title='Parameters', fontsize=10)

# Subplot 2: Varying Frequency
ax2.set_title('Varying Frequency', fontsize=16)
ax2.set_xlabel('x', fontsize=14)
ax2.set_ylabel('y', fontsize=14)
ax2.legend(title='Parameters', fontsize=10)

# Subplot 3: Varying Both
ax3.set_title('Varying Both Amplitude and Frequency', fontsize=16)
ax3.set_xlabel('x', fontsize=14)
ax3.set_ylabel('y', fontsize=14)
ax3.legend(title='Parameters', fontsize=10)

plt.show()
```
