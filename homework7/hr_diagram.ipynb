```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
print(f"Shape of the data (rows, columns): {data.shape}")
# 2642 rows (stars)
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
min_mag = np.min(magnitude)
max_mag = np.max(magnitude)

print(f"Minimum Magnitude: {min_mag}")
print(f"Maximum Magnitude: {max_mag}")

idx_brightest = np.argmin(magnitude) 
idx_dimmest = np.argmax(magnitude)

print(f"Minimum Magnitude: {idx_brightest}")
print(f"Maximum Magnitude: {idx_dimmest}")
```

```python
min_temp = np.min(temperature)
max_temp = np.max(temperature)

print(f"Minimum Temperature (Coldest star): {min_temp} K")
print(f"Maximum Temperature (Hottest star): {max_temp} K")

idx_coldest = np.argmin(temperature)
idx_hottest = np.argmax(temperature)

print(f"Minimum Temperature (Coldest star): {idx_coldest}")
print(f"Maximum Temperature (Hottest star): {idx_hottest}")
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, label='Stars', c='black', marker='1')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.gca().invert_xaxis() and plt.gca().invert_yaxis()

plt.scatter(log_temperature, log_magnitude, label='Stars', c='black', marker='1')

plt.scatter(log_temperature[idx_brightest], log_magnitude[idx_brightest],  label='Brightest Star', color='black', marker='*', s=150)

plt.scatter(log_temperature[idx_dimmest], log_magnitude[idx_dimmest],  label='Dimmest Star', color='purple', marker='*', s=150)

plt.scatter(log_temperature[idx_hottest], log_magnitude[idx_hottest], label='Hottest Star', color='green', marker='*', s=150)

plt.scatter(log_temperature[idx_coldest], log_magnitude[idx_coldest], label='Coldest Star', color='orange', marker='*', s=150)

plt.title('Hertzsprung-Russel Diagram', fontsize=16)
plt.xlabel('Temperature (log K)', fontsize=12)
plt.ylabel('Magnitude', fontsize=12)
plt.legend(loc='lower left', fontsize=10)
plt.legend()

plt.show()
```
